# PRESSURE-DRIVEN FLOW IN A SLOT WITH A MOVING WALL

A slot is a 2D channel with rigid parallel walls that extend infinitely out of the page.

The dimensionless solution for the combined Poiseuille&ndash;Couette flow in a slot is
$$
    \frac{v_x}{V_0}=v_x^*(y^*)=y^* + \frac{1}{2}\mathbb{P}\mathrm{o}[y^* - (y^*)^2],
    \qquad y^*=\frac{y}{h},
$$
as given in Sections 7.4 and 8.10 of Panton.

Here,
$$\mathbb{P}\mathrm{o} = \frac{h^2\Delta p}{\mu_0 L V_0} = \frac{\Delta p/L}{\mu_0 V_0/h^2} = \frac{\text{imposed pressure gradient}}{\text{force (per vol.) to drive the top wall}}$$
is a dimensionless number that I've decided to call the _Poiseuille number_ (but be careful that other [dimensionless numbers](https://en.wikipedia.org/wiki/Dimensionless_numbers_in_fluid_mechanics) may also carry this name).

$\mathbb{P}\mathrm{o}$ can be positive or negative, depending on whether the pressure gradient is favorable or adverse.

[ASIDE: Note that we interpret $\mu_0 V_0/h^2$ as the force (per vol.) because the viscous term balacing the pressure gradient in the $x$-momentum equation is $\mu_0 \partial^2 v_x/\partial y^2$ which can be made dimensionless by divinging by precisely $\mu_0 V_0/h^2$.]

I will use a [`quiver`](https://matplotlib.org/stable/gallery/images_contours_and_fields/quiver_simple_demo.html) plot to show the velocity arrows, and a regular line plot to show he envelope of the vectors.

```python
def plot_func(Po, save=False):
    # y^* goes from 0 to 1
    y = np.linspace(0, 1)
    # v_x^* as defined above
    vx = y + 0.5*Po*(y - y**2)

    fig, ax = plt.subplots(tight_layout=True)
    ax.set_ylabel('$y^* = y/h$')
    ax.set_ylim(0, 1)
    ax.set_xlabel('$v_x^* = v_x/V_0$')

    # we have to be judicious with the xlim interval,
    # to make sure we can see the flow for all values of Po
    # allowed in the slider
    ax.set_xlim(-2.5, 3.5)

    # skip every other data point (using "::2") to reduce number of arrows
    # plot arrows emanating only from the reference line, x-coord = 0
    # also had to hack the labe font a little, and .2f rounds to 2 decimals
    ax.plot(vx, y, color='darkblue', linewidth=4,
            label=rf"$\mathrm{{\mathbb{{P}}}}\mathrm{{o}}={Po:.2f}$")
    ax.quiver(0*y[::2], y[::2],
              vx[::2], 0*vx[::2],
              scale=1., scale_units='xy', color='blue', zorder=2)

    ax.grid(alpha=0.5, linestyle='dotted')

    ax.legend(loc='lower right')

    if save:
        fig.savefig(save) # bbox_inches='tight'

    plt.show()
```

First, a static plot.

```python
plot_func(1)
```

Then, an [`interact`](https://ipywidgets.readthedocs.io/en/latest/examples/Using%20Interact.html) widget will let us vary $\mathbb{P}\mathrm{o}$.

```python
interact(plot_func, 
         Po = widgets.FloatSlider(value=1, min=-20, max=+20, step=0.5, description='Pois. #, Po'),
         save=fixed(False));
```

Playing with the slider on the plot, we see that changing the $\mathbb{P}\mathrm{o}$ number changes the shape of the flow profile.

# ANOTHER REPRESENTATION OF THE SAME SOLUTION

Observing that $h^2\Delta p/(\mu_0L)$ is also a _velocity_ (in fact, it is $12\times$ the average velocity of the pressure-driven (Poiseuille) flow, see also Panton p. 131), we could have written the dimensionless solution as
$$
    \frac{v_x}{h^2\Delta p/(\mu_0L)}=v_x^*(y^*)= \widetilde{\mathbb{P}{\mathrm{o}}} \, y^*  + \frac{1}{2}[y^* - (y^*)^2],\qquad y^*=\frac{y}{h}.
$$

Here,
$$\widetilde{\mathbb{P}\mathrm{o}} = \frac{\mu_0 L V_0}{h^2\Delta p} = \frac{V_0}{h^2\Delta p/(\mu_0 L)} = \frac{\text{wall velocity}}{12\times\text{avg velocity of the pressure driven flow}} = \frac{1}{\mathbb{P}\mathrm{o}}$$
is the (inverse of the) _Poiseuille number_ from before.

Both, $\mathbb{P}\mathrm{o}$ and $\widetilde{\mathbb{P}\mathrm{o}}=1/\mathbb{P}\mathrm{o}$ , are equally valid dimensionless number to describe this flow.

```python
def plot_func_tilde(PoTilde, save=False):
    # y^* goes from 0 to 1
    y = np.linspace(0, 1)
    # v_x^* as defined above
    vx = PoTilde*y + 0.5*(y - y**2)

    fig, ax = plt.subplots(tight_layout=True)
    ax.set_ylabel('$y^* = y/h$')
    ax.set_ylim(0, 1)
    ax.set_xlabel(r'$v_x^* = \frac{v_x}{h^2\Delta p/(\mu_0L)}$')
    ax.set_xlim(-2, 2)

    # skip every other data point (using "::2") to reduce number of arrows
    # plot arrows emanating only from the reference line, x-coord = 0
    # also had to hack the labe font a little, and .2f rounds to 2 decimals
    ax.plot(vx, y, color='darkblue', linewidth=4,
            label=rf"$\widetilde{{\mathrm{{\mathbb{{P}}}}\mathrm{{o}}}}={PoTilde:.2f}$")

    ax.quiver(0*y[::2], y[::2],
              vx[::2], 0*vx[::2],
              scale=1., scale_units='xy', color='blue', zorder=2)

    ax.grid(alpha=0.5, linestyle='dotted')

    ax.legend(loc='lower right')

    if save:
        fig.savefig(save) # bbox_inches='tight'

    plt.show()
```

```python
plot_func_tilde(1)
```

```python
interact(plot_func_tilde, 
         PoTilde=widgets.FloatSlider(value=1, min=-2.5, max=+2.5, step=0.05, description='Pois. tilde #'),
         save=fixed(False));
```
